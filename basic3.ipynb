```python
import matplotlib.pyplot as plt
```

```python
fig,axes = plt.subplots(    #subplots는 tuple을 반환,파이썬:하나의 tuple에 여러가지 변수형 반환
    nrows=1,    #한 줄에 2개여서 행은 하나,열은 두개 / 표 하나하나는 figure라고 한다. figsize=표 크기
    ncols=2,
    figsize=(10,4),
)

plt.show()   # 비어있는 축 2개가 그려짐
```

```python
fig,axes = plt.subplots(
    nrows=2,
    ncols=2,
    figsize=(8,6),
)

print(axes)   # [[<Axes: > <Axes: >] 배열로 들어있다,axes로 특정그래프,축 불러올수있다.
print(axes[0,0])

plt.show()
```

```python
fig,ax = plt.subplots(   # nrow,ncol 지정 안해서 축 하나임>axes말고 ax
    figsize=(8,4),
)

x = [1,2,3,4,5]
y = [10,15,13,20,24]

ax.plot(
    x,
    y,
    linestyle='--', #dashed='--'그래프 점선으로
    linewidth=2,
    marker='o', #마커 스타일 세모로 marker='^'
    markersize=8
)

plt.show()
```

```python
fig,ax = plt.subplots(figsize=(8,4))

month = [1,2,3,4,5,6]

product_a=[120,130,128,160,175,190]
product_b=[100,110,125,140,150,165]

ax.plot(month,product_a,marker='o',label='Product A')
ax.plot(month,product_b,marker='s',label='Product B')

ax.legend()  #artist 객체가 뭔지? matplotlib은 그렇게 깊게 안파도 ㄱㅊ

plt.show()
```
