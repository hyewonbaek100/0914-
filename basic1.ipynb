```python
df=pd.DataFrame({
    'month':[1,2,3,4,5,6],
    'sales':[120,135,128,160,175,190]
})

#col=df['sales']
#type(col)

plt.plot(df['month'],df['sales'])
plt.show()
# 배열도 그래프로 그릴 수 있구나! 대신 x축,y축에 어떤거넣을지
```

```python
month=[1, 2, 3, 4, 5, 6]

product_a=[120,135,128,160,175,190]
product_b=[90,110,125,120,150,165]
#한 줄 복사할때 ctrl+d

#같은 축을 공유(x)하고 값(y)이 다른 선을 두 개 그려야한다.

plt.plot(month,product_a)
plt.plot(month,product_b)

plt.show()
```

```python
# 범례 : 뭐가 뭘 나타내는지 표시를 해준다.label(레이블)
month=[1, 2, 3, 4, 5, 6]

product_a=[120,135,128,160,175,190]
product_b=[90,110,125,120,150,165]


plt.plot(month,product_a,label='Product A')
plt.plot(month,product_b,label='Product B')

plt.legend()  #부여한 범례를 시각화하기 위함

plt.show()
```

```python
month=[1, 2, 3, 4, 5, 6]

product_a=[120,135,128,160,175,190]
product_b=[90,110,125,120,150,165]
product_c=[70,85,95,110,125,140]


plt.plot(month,product_a,label='Product A')
plt.plot(month,product_b,label='Product B')
plt.plot(month,product_c,label='Product C')

plt.legend()  #부여한 범례를 시각화하기 위함

plt.show() #하나의 그래프에는 비교 가능한 것만 집어넣는 게 좋다,많은 것들을 넣으면 비교불가
```

```python
month=[1, 2, 3, 4, 5, 6]

product_a=[120,135,128,160,175,190]

# title='monthly sales'
title='Monthly Sales'

plt.plot(month,product_a)
plt.title(
    title,
    fontsize=18,
   # loc='left',
)

plt.xlabel('Month')
plt.ylabel('Sales')

print(plt.gca().get_title()) #그래프의 제목을 문자열로 가지고 올때
# print(type(plt.gca()))  #제목 정보도 축 정보 안에 있다

print(plt.gca().get_xlable())
print(plt.gca().get_ylable())

plt.show()
```

```python
# 툭정 축에 대한 범위를 제한/지정하고 싶을 때 lim()

month=[1, 2, 3, 4, 5, 6]

product_a=[120,135,128,160,175,190]

plt.plot(month,product_a,marker='o')  #marker=o 각각의 point를 o로 보여줌
plt.xlim(0,12)
plt.ylim(90,200)
plt.show()
```

```python
month=[1, 2, 3, 4, 5, 6]

product_a=[120,135,128,160,175,190]

plt.plot(month,product_a)

plt.xticks( #눈금을 1~6까지 직접 지정한 것
    [1,2,3,4]  ,
    ['January','February','March','April'],
    rotation=45,
)
# xticks에서 눈금에 대한 이름을 줄 때
# 눈금에 대한 이름이 길어졌을 때 기울이는거=rotation

plt.show()
```
