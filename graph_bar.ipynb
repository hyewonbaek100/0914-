```python
import matplotlib.pyplot as plt
```

```python
products = ['A','B','C','D']   #products가 x축에 오는걸 알아야한다,x축=가로에 오는것
sales = [120,180,150,210]

fig,ax = plt.subplots(figsize=(10,4))

ax.bar(products,sales)

plt.show()

```

```python
products = ['Product A','Product B','Product C','Product D']
sales = [120,180,150,210]

fig,ax = plt.subplots(figsize=(8,4))

ax.barh(products,sales) #ax.barh>h=수평선 의미 #보여주는 것만 수평이어서 x가 y축으로 간다.
plt.show()
```

```python
products = ['Product A','Product B','Product C','Product D']
sales = [120,180,150,210]

fig,ax = plt.subplots(figsize=(8,4))

ax.barh(products,sales)

ax.set_title('Products Sales')
ax.set_xlabel('Sales')   #그림에서는 y축이어도 데이터가 x축이면 알아서 뒤집는다
ax.set_ylabel('Products')

plt.show()
```

```python
products = ['Product A','Product B','Product C','Product D']
sales = [120,180,150,210]

fig,ax = plt.subplots(figsize=(10,4))   #가로길이,세로길이

bars = ax.barh(products,sales)

ax.bar_label(
    bars,     #해당되는 축에 대한 값을 띄워준다
    padding=3    #안쪽으로 3픽셀만큼 늘어난거다 막대바와 값을 붙이고 띄워주는 역할
)

ax.set_title('Products Sales')
ax.set_xlabel('Sales')
ax.set_ylabel('Products')

plt.show()
```

```python
products = [ 'A','B','C']

sales_2026=[120,160,140]
sales_2025=[150,180,170]

x = np.arange(len(products))  # 이렇게 해야 0,1,2배열나온다
width=0.35

fig,ax = plt.subplots(figsize=(8,4))

ax.bar(
    x - width/2,
    sales_2025,
    width,
    label='2025'
)

ax.bar (
    x + width/2,
    sales_2026,
    width,
    label='2026'
)

ax.set_xticks(x)
ax.set_xticklabels(products)

ax.set_title('Sales by year')
ax.set_xlabel('products')
ax.set_ylabel('sales')

ax.legend()

plt.show()   # 하나의 축에 여러 바를 두고 비교가 가능하다
```

```python
products = [ 'A','B','C']

q1=[120,150,130]
q2=[140,170,145]
q3=[160,180,155]

width=0.25

x = np.arange(len(products))
width=0.25

fig,ax = plt.subplots(figsize=(10,4))

ax.bar(
    x - width,
    q1,
    width=width,
    label='Q1'
)

ax.bar (
    x ,
    q2,
    width=width,
    label='Q2'
)

ax.bar (
    x + width,
    q3,
    width=width,
    label='Q3'
)

ax.set_xticks(x)
ax.set_xticklabels(products)

ax.legend()

plt.show()
```

```python
products = ['A', 'B', 'C']

online = [100, 130, 120]
offline= [50, 70, 80]

fig,ax= plt.subplots(figsize=(8,4))

ax.bar(
    products,
    online,
    label='Online'
)

ax.bar(
    products,
    offline,
    bottom=online,
    label='Offline'
)

ax.bar(
    products,
    [10,20,30],
    label='Others'
)

ax.legend()

plt.show()
```
