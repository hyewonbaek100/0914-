```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sales_df = pd.DataFrame({
    "month": [
        1, 2, 3, 4, 5, 6,
        1, 2, 3, 4, 5, 6,
        1, 2, 3, 4, 5, 6,
    ],
    "region": [
        "Seoul", "Seoul", "Seoul",
        "Seoul", "Seoul", "Seoul",
        "Busan", "Busan", "Busan",
        "Busan", "Busan", "Busan",
        "Daejeon", "Daejeon", "Daejeon",
        "Daejeon", "Daejeon", "Daejeon",
    ],
    "sales": [
        120, 138, 155, 171, 188, 205,
        105, 118, 132, 149, 163, 178,
        92, 106, 119, 133, 148, 162,
    ],
    "advertising": [
        10, 12, 15, 18, 21, 24,
        9, 11, 13, 16, 19, 22,
        7, 9, 12, 14, 17, 20,
    ],
    "visitors": [
        800, 900, 1010, 1130, 1260, 1400,
        720, 810, 900, 1010, 1120, 1250,
        650, 730, 820, 910, 1020, 1140,
    ],
})

sales_df
```

## 테마 세팅

```python
# 전엽적인 환경 = global 지엽적인 설정 = local
# seaborn 에 직접적으로 테마를 읽혀줄거다, 코드 실행 이후에는 해당 파일에 대해 global하게 적용된다
```

```python
sns.set_theme()  #기본 테마 입혀짐
```

```python
sns.lineplot(
    data=sales_df,
    x="month",
    y="sales",
    hue="region",
    marker='o'
)
```

```python
sns.set_theme(style='darkgrid')   #seaborn에 테마 적용하면 멧플랏립의 데이터에도 테마가 적용된다

plt.plot(
    [1,2,3,4],
    [10,15,13,20],
    marker='o'
)

plt.show()
```

```python
sns.set_theme(
    style='whitegrid',
    context='notebook',    # 책에서 볼 수 있는 것처럼 적정한 폰트 크기로 적용된다
    palette='deep'           # 팔렛트=색감 , deep=짙은
)
```

```python
sns.lineplot(
    data=sales_df,
    x="month",
    y="sales",
    hue="region",
    marker='o'
)
```

# Color Palette

1. 서로 다른 범주 구분
   - 범주형 팔레트    eg.서울,대전,부산
2. 작은 값에서 큰 값으로 변화 표현
   - 연속형 팔레트
3. 특정 기준보다 낮거나 높은 방향 표현
   - 발산형 팔레트

```python
palette= sns.color_palette(
    'deep',
    n_colors=6,     #6개의 색상 코드만 꺼내온다
)
```

```python
sns.color_palette(
    'pastel',
    6
)
```

```python
p = sns.color_palette(
    'muted',
    6
)

sns.palplot(p)   #palplot > 그림으로 나타내주는
```

```python
sns.set_theme(
    style='whitegrid',
    context='notebook',
    palette='pastel'
)

sns.lineplot(
    data=sales_df,
    x='month',
    y='sales',
    hue='region',
    marker='o'
)
```

```python
sns.scatterplot(
    data=sales_df,
    x='month',
    y='sales',
    hue='region',
    palette='Set2',
    s = 80
)
```
