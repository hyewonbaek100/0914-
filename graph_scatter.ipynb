```python
import matplotlib.pyplot as plt
import pandas as pd

df =pd.DataFrame({
    'advertising':[10,15,20,25,30,35,40,45],
    'sales':[100,115,125,145,150,175,185,195]
})

fig,ax=plt.subplots(figsize=(8,6))

ax.scatter(
    df['advertising'],
    df['sales'],
    color='orange',
    s=150
)

ax.set_title('Advertising and Sales')
ax.set_xlabel('Advertising')
ax.set_ylabel('Sales')

plt.show()
```

```python
df=pd.DataFrame({
    'study_hours' : [2,3,4,5,6,2.5,4.5,5.5,7],
    'scores' : [60,67,72,80,86,58,75,82,91],
    'class' : ['A','A','A','A','A','B','B','B','B']
})

fig,ax=plt.subplots(figsize=(8,6))

class_a=df[df['class']=='A']
class_b=df[df['class']=='B']

ax.scatter(
    class_a['study_hours'],
    class_a['scores'],
    label = 'Class A'
)

ax.scatter(
    class_b['study_hours'],
    class_b['scores'],
    label = 'Class B'
)

ax.set_xlabel('Study Hours')
ax.set_ylabel('Score')

ax.legend()

plt.show()
```

```python
df=pd.DataFrame({
    'x':[1,2,3,2,3,4,3,4,5],
    'y':[5,7,8,4,6,9,3,7,10],
        'group': [
        'A','A','A',
        'B','B','B',
        'C','C','C',
    ]
})

fig,ax = plt.subplots(figsize=(8,6))

# group_a=df[df['group']=='A']   #그룹별 필터링

  #ax.scatter(
   # group_a['x'],
    #group_a['y'],
    #label='Group A',
    #s=80
#)

for group_name in ['A','B','C']:
    group=df[df['group']==group_name]

    ax.scatter(
        group['x'],
        group['y'],
        label=f'Group{group_name}',
        s=80
    )

ax.legend()

plt.show()
```

```python
import numpy as np

rng = np.random.default_rng(42)

x = rng.normal(
    loc = 50,
    scale=10,
    size = 500,
)

y = ( x * 0.7 + rng.normal(loc=0,scale=8,size=500))

print(len(x),len(y))
```

```python
import matplotlib.pyplot as plt

fig,ax=plt.subplots(figsize=(8,6))

ax.scatter(x,y,s=40,alpha=0.25)  #alpha는 0~1 값을 가짐

plt.show()
```

```python
rng = np.random.default_rng(42)  # 난수 생성기

scores=rng.normal(70,10,100) #100개의 난수가 생성됐다

fig,ax = plt.subplots(figsize=(8,6))

ax.hist(scores)  #히스토그램 생성

plt.show()
```
